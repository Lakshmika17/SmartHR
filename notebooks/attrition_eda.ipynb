```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)

BASE_DIR = Path(r"E:\Projects\SmartHR")
DATA_DIR = BASE_DIR / "Datasets"

print("Libraries imported successfully!")
```

```python

file_path = DATA_DIR / "WA_Fn-UseC_-HR-Employee-Attrition.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe())
```

```python

print("Missing values in each column:")
display(df.isnull().sum().sort_values(ascending=False))

print("\nTotal missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

print("\nAttrition distribution:")
display(df["Attrition"].value_counts())

print("\nAttrition percentage:")
display(
    (df["Attrition"].value_counts(normalize=True) * 100).round(2)
)
```

```python

cleaned = df.copy()

# Remove duplicate rows
cleaned = cleaned.drop_duplicates()

# Identify constant columns
constant_columns = [
    col for col in cleaned.columns
    if cleaned[col].nunique(dropna=False) <= 1
]

print("Constant columns to remove:", constant_columns)

# Remove columns with no variation
cleaned = cleaned.drop(columns=constant_columns)

# Check missing values after cleaning
print("\nMissing values after cleaning:",
      cleaned.isnull().sum().sum())

print("Cleaned dataset shape:", cleaned.shape)

# Save cleaned dataset
output_path = DATA_DIR / "attrition_cleaned.csv"
cleaned.to_csv(output_path, index=False)

print("\nCleaned dataset saved at:", output_path)

display(cleaned.head())
```

```python

plt.figure(figsize=(6, 4))

sns.countplot(data=df, x="Attrition")

plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Department", hue="Attrition")

plt.title("Employee Attrition by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Age",
    hue="Attrition",
    multiple="stack",
    bins=15
)

plt.title("Age Distribution by Attrition")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.tight_layout()
plt.show()
```

```python

income = df.groupby("Attrition")["MonthlyIncome"].mean()

print("Average monthly income by attrition:")
display(income.round(2))

plt.figure(figsize=(6, 4))
sns.barplot(x=income.index, y=income.values)

plt.title("Average Monthly Income by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Average Monthly Income")
plt.tight_layout()
plt.show()
```

```python

years = df.groupby("Attrition")["YearsAtCompany"].mean()

print("Average years at company:")
display(years.round(2))

plt.figure(figsize=(6, 4))
sns.barplot(x=years.index, y=years.values)

plt.title("Average Years at Company by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Average Years")
plt.tight_layout()
plt.show()
```

```python

overtime_table = pd.crosstab(
    df["OverTime"],
    df["Attrition"],
    normalize="index"
) * 100

print("Attrition percentage by overtime status:")
display(overtime_table.round(2))

overtime_table.plot(kind="bar", figsize=(7, 4))

plt.title("Attrition Percentage by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Percentage (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

numeric_df = df.select_dtypes(include="number")

plt.figure(figsize=(14, 9))

sns.heatmap(
    numeric_df.corr(),
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap - Employee Attrition")
plt.tight_layout()
plt.show()
```
